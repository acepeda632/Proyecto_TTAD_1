```python
import json 
with open('data.json', 'r', encoding='utf-8') as file:
    data = json.load(file  )

ventas = data['ventas']
print(len(ventas), "ventas cargadas")
```

```python
ingresos = {}
for venta in ventas:
    producto = venta['producto']
    importe = venta['cantidad'] * venta['precio_unitario']
    if producto in ingresos:
        ingresos[producto] += importe
    else:
        ingresos[producto] = importe

mejor_producto = ""
mayor_ingreso = 0
for producto in ingresos:
    if ingresos[producto] > mayor_ingreso:
        mejor_producto = producto
        mayor_ingreso = ingresos[producto]

print("Producto con más ingresos:", mejor_producto, "con", f"{mayor_ingreso:.2f} €")
print(ingresos)
```

```python
vendedor = "Ana"

ventas_vendedor = []
total_vendedor = 0
for venta in ventas:
    if venta['vendedor'] == vendedor:
        ventas_vendedor.append(venta)
        total_vendedor += venta['cantidad'] * venta['precio_unitario']

print("Ventas de", vendedor, ":", len(ventas_vendedor))
print("Total facturado por", vendedor, ":", f"{total_vendedor:.2f} €")
for venta in ventas_vendedor:
    print(venta)
```

```python
total = 0
for venta in ventas:
    total += venta['cantidad'] * venta['precio_unitario']

ticket_medio = total / len(ventas)
print("Ticket medio:", f"{ticket_medio:.2f} €")
```
